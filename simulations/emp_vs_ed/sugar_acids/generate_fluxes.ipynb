```python
import cobra
import json
import pickle as pkl
from gem_utilities import media
```

```python
# Load the model
model = cobra.io.read_sbml_model("../../../model.xml")
```

```python
# Load the media definitions
with open("../../../test/test_files/media/media_definitions.pkl", "rb") as f:
    media_definitions = pkl.load(f)
mbm = media_definitions["mbm"]
```

```python
# Define carbon sources to test
c_sources = {
    "galactose": "cpd00108_e0",
    "mannuronate": "cpd01384_e0",
    "galacturonate": "cpd00280_e0"
}
```

```python
# For each carbon source, set the model to use that carbon source and run FBA
for c_source, met_id in c_sources.items():
    # Get the number of carbons in the source from the model
    met = model.metabolites.get_by_id(met_id)
    num_carbons = met.elements.get("C", 0)
    # Get the exchange reaction for the carbon source
    reaction_id = f"EX_{met_id}"
    # Set the carbon source as the only carbon source in the media
    new_medium = mbm.copy()
    new_medium[reaction_id] = 60 / num_carbons  # Set uptake rate to 60 mmol/gDW/hr normalized by carbons
    model.medium = media.clean_media(model, new_medium)
    # Run pFBA
    solution = cobra.flux_analysis.pfba(model)
    # Save the fluxes as a JSON file
    with open(f"results/fluxes_{c_source}.json", "w") as f:
        json.dump(solution.fluxes.to_dict(), f)
```
