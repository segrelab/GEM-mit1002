```python
import cobra
import pandas as pd
import json
```

```python
# Load the model
model = cobra.io.read_sbml_model("/Users/helenscott/Documents/PhD/Segre-lab/GEM-repos/GEM-mit1002/model.xml")
```

```python
# Load the ko results
ko_results = pd.read_csv("results/single_reaction_ko_results.csv", index_col=0)
```

```python
ko_results
```

```python
# Filter the data to only include reactions that are essential in at least one condition (growth rate < minimum growth rate threshold)
essential_reactions_df = ko_results[ko_results.lt(0.001).any(axis=1)]
essential_reactions_df
```

```python
# Are there any reactions that are essential in only some conditions?
essential_in_less_than_all_conditions = essential_reactions_df[essential_reactions_df.gt(0.001).any(axis=1)]
essential_in_less_than_all_conditions
```

```python
# Load my escher map json file
escher_map = json.load(open("/Users/helenscott/Documents/PhD/Segre-lab/GEM-repos/GEM-mit1002/escher/MIT1002_glycolysis-tca-etc_escher-map.json"))
```

```python
# Get a list of all the reactions in the escher map
rxns_in_map = [escher_map[1]["reactions"][r]["bigg_id"] for r in escher_map[1]["reactions"]]
```

```python
# What essential reactions are in the escher map?
essential_in_map = essential_reactions_df[essential_reactions_df.index.isin(rxns_in_map)].index
for rxn in essential_in_map:
    print(rxn , ": ", model.reactions.get_by_id(rxn).build_reaction_string(use_metabolite_names=True))
```

```python
# What essential reactions are NOT the escher map?
essential_not_in_map = essential_reactions_df[~essential_reactions_df.index.isin(rxns_in_map)].index
for rxn in essential_not_in_map:
    print(rxn , ": ", model.reactions.get_by_id(rxn).build_reaction_string(use_metabolite_names=True))
```

```python
# Add a reaction name column to the essential reactions dataframe
essential_reactions_df["reaction_name"] = essential_reactions_df.index.map(lambda x: model.reactions.get_by_id(x).name)
# Add a reaction string column to the essential reactions dataframe
essential_reactions_df["reaction_string"] = essential_reactions_df.index.map(lambda x: model.reactions.get_by_id(x).build_reaction_string(use_metabolite_names=True))
# Save
essential_reactions_df.to_csv("results/essential_reactions_with_names_and_strings.csv")
```
