```python
import cobra
```

```python
model = cobra.io.read_sbml_model("model.xml")
```

```python
# Get the glutamine metaboltie
glut_met = model.metabolites.get_by_id("cpd00053_c0")
glut_met
```

```python
# Define an exometabolite for glutamine
glut_exo = cobra.Metabolite(
    id="cpd00053_e0",
    name="L-Glutamine [e0]",
    compartment="e0",
    formula=glut_met.formula,
    charge=glut_met.charge
)
glut_exo
```

```python
# Look up the glutamate transport reaction in the model
model.reactions.get_by_id("rxn05298_c0")
```

```python
# Define a transport reaction for glutamine
trans_rxn = cobra.Reaction(
    "rxn05216_c0",
    name="Na+/glutamine symport [c0]",
    lower_bound=-1000.0,)
```

```python
trans_rxn.add_metabolites({
    glut_exo: -1.0,
    model.metabolites.cpd00971_e0: -1.0,
    glut_met: 1.0,
    model.metabolites.cpd00971_c0: 1.0
})
```

```python
trans_rxn
```

```python
# Add to the model
model.add_reactions([trans_rxn])
```

```python
# Add exchange reaction for glutamine
model.add_boundary(model.metabolites.cpd00053_e0, type="exchange", lb=0, ub=1000)
```

```python
# Save the model
cobra.io.write_sbml_model(model, "model.xml")
```
