```python
import cobra
```

```python
model = cobra.io.read_sbml_model("model.xml")
```

```python
# Get the relevant metabolites
h_c    = model.metabolites.get_by_id("cpd00067_c0")   # cytosolic H+
h_e    = model.metabolites.get_by_id("cpd00067_e0")   # extracellular H+
pg6_c  = model.metabolites.get_by_id("cpd00284_c0")   # cytosolic 6-P-gluconate
```

```python
# Add extracellular 6PG
pg6_e = cobra.Metabolite(
    "cpd00284_e0",
    name="6-Phospho-D-gluconate",
    formula="C6H10O10P",   # matches ModelSEED cpd00284 → keeps balance
    charge=-3,
    compartment="e0",
)
model.add_metabolites([pg6_e])
```

```python
# Add the transport reaction
t = cobra.Reaction("rxn05482_c0")
t.name = "6-Phospho-D-gluconate transport via proton symport"
t.lower_bound = -1000.0
t.upper_bound = 1000.0
t.add_metabolites({
    h_c:   -1.0,
    pg6_c: -1.0,
    h_e:    1.0,
    pg6_e:  1.0,
})
model.add_reactions([t])
```

```python
# Add the exchange reaction for extracellular 6PG
model.add_boundary(pg6_e, type="exchange", reaction_id="EX_cpd00284_e0", lb=0)
```

```python
# Save the model
cobra.io.write_sbml_model(model, "model.xml")
cobra.io.save_json_model(model, "/Users/helenscott/Desktop/model.json")
```
